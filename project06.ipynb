## Background

Today we will implement a distance-based phylogenetic tree construction method using the Neighbor-Joining (NJ) algorithm and Smith-Waterman local alignment scores. Unlike UPGMA (Unweighted Pair Group Method using Arithmetic averages), NJ produces unrooted trees and does not assume a constant evolutionary rate across lineages, making it more biologically realistic for analyzing sequence relationships.

We will use HIV-1 reverse transcriptase sequences to construct our phylogenetic tree. The Smith-Waterman algorithm will be used to generate pairwise local alignment scores, which will then be converted into distances for tree construction. This approach is particularly suitable for HIV sequence analysis as it:
1. Handles sequence variations effectively through local alignment
2. Accounts for potential rate heterogeneity across different viral strains
3. Does not assume a molecular clock

The ultimate output will be an unrooted phylogenetic tree representing the evolutionary relationships between the HIV-1 sequences, visualized using the ete3 library. This method provides insights into viral diversity and evolutionary patterns while avoiding the assumptions of simpler hierarchical clustering approaches.

The key innovations of this implementation are:
- Use of Smith-Waterman for sensitive local alignment scoring
- Implementation of Neighbor-Joining for unrooted tree construction
- More biologically realistic evolutionary model

## Neighbor-Joining Algorithm

**Input**: Distance matrix $D$ for $n$ sequences

**Initialization**:
- Let $n$ be the number of sequences
- Assign each sequence $i$ to its own leaf node
- Initialize tree $T$ with leaf nodes

**Iteration**:
While $n > 2$:
1. Calculate Q-matrix where:
   $Q(i,j) = (n-2)d(i,j) - \sum_{k=1}^n d(i,k) - \sum_{k=1}^n d(j,k)$

2. Find pair $(i,j)$ with minimum $Q(i,j)$

3. Calculate branch lengths:
   $d_i = \frac{d(i,j) + (r_i - r_j)/(n-2)}{2}$
   $d_j = d(i,j) - d_i$
   where $r_i = \sum_{k=1}^n d(i,k)$

4. Create new node $k$
   - Add branches from $k$ to $i$ and $j$ with lengths $d_i$ and $d_j$

5. Update distances to remaining nodes $x$:
   $d(k,x) = \frac{d(i,x) + d(j,x) - d(i,j)}{2}$

6. Remove nodes $i$ and $j$
7. Add node $k$ to active nodes
8. $n = n - 1$

**Termination**:
When $n = 2$ with remaining nodes $i$ and $j$:
- Add final branch between $i$ and $j$ with length $d(i,j)$
- Return unrooted tree $T$

````python
from Bio import SeqIO
def read_fasta(filename: str) -> Dict[str, str]:
    """Reads sequences from FASTA file.

    Args:
        filename (str): Path to FASTA file containing HIV RT sequences

    Returns:
        Dict[str, str]: Dictionary mapping sequence IDs to sequences

    Examples:
        >>> seqs = read_fasta("lafayette_SARS_RT.fasta")
        >>> len(seqs) > 0
        True
    """

#use the last word of the full header as the ID
#record.id would give the first word and the full header contains commas and brackets that would break the Newick string
#str() converts biopython Seq object to a plain string for smith_waterman


    fasta_dict = {record.description.split()[-1]: str(record.seq)
        for record in SeqIO.parse(filename, "fasta")}
    return fasta_dict
````

````python

def cal_score(matrix, seq1, seq2, i, j, match, mismatch, gap):
    """Calculate score for position (i,j) in scoring matrix,
    
        Args:
            matrix (numpy array): scoring matrix
            seq1 (str): sequence 1
            seq2 (str): sequence 2
            i (int): current row number
            j (int): current column number
        
        Returns:
            score in position (i,j)    
            move to trace back: 0-END, 1-DIAG, 2-UP, 3-LEFT
    
        """
    
    #the matrix has an extra row and column of zeros at the start, so row i corresponds to seq1[i - 1] and column j to seq2[j - 1]

    if seq1[i - 1] == seq2[j - 1]:
        diag_score = matrix[i - 1][j - 1] + match #if the characters align and match
    else:
        diag_score = matrix[i - 1][j - 1] + mismatch #if the characters align but differ

    #moving down or right means adding a gap
    up_score = matrix[i - 1][j] + gap
    left_score = matrix[i][j - 1] + gap

    #including 0 for local alignment to prevent negatives from dragging the score down
    best_score = max(0, diag_score, up_score, left_score)

    return best_score
````

````python
def smith_waterman(
    sequence1: str,
    sequence2: str,
    match_score: float = 2.0,
    mismatch_penalty: float = -1.0,
    gap_penalty: float = -1.0
) -> float:
    """Implements Smith-Waterman local alignment algorithm.

    Args:
        sequence1 (str): First HIV RT sequence
        sequence2 (str): Second HIV RT sequence
        match_score (float): Score for matching characters
        mismatch_penalty (float): Penalty for mismatched characters
        gap_penalty (float): Penalty for gaps

    Returns:
        float: Normalized similarity score between sequences

    Examples:
        >>> seq1 = "ACGT"
        >>> seq2 = "ACGT"
        >>> smith_waterman(seq1, seq2)
        1.0

    Pseudocode: 
    raw local alignment score (adapted from project 05, traceback not required here)
    m, n = lengths of sequence1, sequence2
    create (m+1) x (n+1) score matrix of zeros (float)
    highest_score = 0
    for i from 1 to m:
        for j from 1 to n:
            score = cal_score(matrix, sequence1, sequence2, i, j, match_score, mismatch_penalty, gap_penalty)
            matrix[i][j] = score
            if score > highest_score:
                highest_score = score

    self-scores: best possible score for each sequence 
    every position matches itself so length * match_score
    self1 = m * match_score
    self2 = n * match_score


    normalize against the average self-score of this pair
    if either self-score is 0 (empty sequence)
        return 0.0
    return highest_score / ((self1 + self2) / 2)
                
            """
    # initialize sequence lengths m and n
    m, n = len(sequence1), len(sequence2)

    # initialize the base maximum score and positions for i and j
    highest_score = 0

    # initialize scoring matrix H with zeros using numpy, add one because S-W uses an extra row/column
    score_matrix = np.zeros((m+1, n+1))
    
    # filling out scoring matrix H
    for i in range(1, m + 1):
        for j in range(1, n + 1):

            # call cal_score on our variables
            best_score = cal_score(score_matrix, sequence1, sequence2, i, j, match_score, mismatch_penalty, gap_penalty)

            # assign the best score into the matrix
            score_matrix[i][j] = best_score

            # Check scoring matrix H against the maximum score and positioning
            if best_score > highest_score:
                highest_score = best_score
                highest_position = (i, j)

    #self scores: best possible score for each sequence (every position matches itself)
    self1 = m * match_score
    self2 = n * match_score

    #avoid dividing by zero if sequence is empty and breaking the loop
    if self1 == 0 or self2 == 0:
        return 0.0
    
    #normalize against the average self-score of this pair
    return highest_score / ((self1 + self2)/2)
````

````python
def build_distance_matrix(sequences: Dict[str, str]) -> Tuple[np.ndarray, List[str]]:
    """Builds DISTANCE (not similiarity) matrix using Smith-Waterman scores.

    Args:
        sequences (Dict[str, str]): Dictionary of HIV RT sequences

    Returns:
        Tuple[np.ndarray, List[str]]: Distance matrix and list of sequence IDs

    Examples:
        >>> seqs = read_fasta("lafayette_SARS_RT.fasta")
        >>> mat, ids = build_distance_matrix(seqs)
        >>> mat[0:3, 0:3]  # Show 3x3 slice of distance matrix
        array([[0.000, 0.004, 0.012],
              [0.004, 0.000, 0.012],
              [0.012, 0.012, 0.000]])
        >>> len(ids)  # Number of sequences
        19

    pseudocode: 
    ids come from the dictionary's keys
    create a matrix of size nxn where n is the number of ids, filled with zeros
    for i from 0 to n-1
        for j from i+1 to n-1 to only visit each pair exactly once
            seq_i = sequences[ids[i]]
            seq_j = sequences[ids[j]]
            score = smith_waterman(seq_i, seq_j)
            keep score between 0.000001 and 1 to avoid -ln(0) = infinity or negative distance scores
            distance = -ln(score)
            matrix[i][j] = distance
            matrix[j][i] = distance
    return matrix, ids

    """

    #puts ids in a fixed order
    ids = list(sequences.keys())
    n = len(ids)
    
    #start with all zeros so sequence self-scores start as 0
    matrix = np.zeros((n, n))

    #only visit pairs where j > i since matrix is symmetric (A->C equals C->A) each pair is aligned once and written to both cells
    for i in range(0,n):
        for j in range(i+1,n):
            seq_i = sequences[ids[i]]
            seq_j = sequences[ids[j]]

            #normalized similarity between 0 and 1
            score = smith_waterman(seq_i, seq_j)

            #keep it between 0.000001 and 1 because -ln(0) is infinite; a score above 1 would give a negative distance
            score = min(max(score, 0.000001), 1.0) 

            #convert similarity to distance using -ln()
            #adding 0.0 turns -0.0 to a clean 0.0
            distance = -np.log(score) + 0.0 

            #fill both matching cells with the one calculation
            matrix[i][j] = distance
            matrix[j][i] = distance
    return matrix, ids
        
````

````python
def neighbor_joining(
    distance_matrix: np.ndarray,
    labels: List[str]
) -> str:
    """Implements Neighbor-Joining algorithm for phylogenetic tree construction.

    Args:
        distance_matrix (np.ndarray): Distance matrix from Smith-Waterman scores
        labels (List[str]): Sequence identifiers

    Returns:
        str: Newick format tree string

    Examples:
        >>> seqs = read_fasta("lafayette_SARS_RT.fasta")
        >>> mat, ids = build_distance_matrix(seqs)
        >>> tree = neighbor_joining(mat, ids)
        >>> tree.startswith("((DM1:0.002,DM2:0.002)")  # Start of Newick string
        True
        >>> tree.count(",")  # Number of separators in tree
        18

    Pseudocode
    make a copy of the distance matrix
    make a copy of the labels list
    n = number of labels
    while n > 2:
        for each row i:
            r[i] = add together all values in row i
        for each pair (i, j) where i < j:
                D*[i][j] = (n-2) * D[i][j] - r[i] - r[j]
        find the pair (i,j) with the smallest D*

        arm_i = (D[i][j] + (r[i] - r[j]) / (n-2)) / 2
        arm_j = D[i][j] - arm_i
        if arm_i < 0:
            arm_i = 0
            arm_j = D[i][j]
        if arm_j < 0:
            arm_j = 0
            arm_i = D[i][j]
        for each node x that is not i or j:
            distance from new junction to x = (D[i][x] + D[j][x] - D[i][j]) / 2
        new_label = f"({labels[i]}:{arm_i},{labels[j]}:{arm_j})"
        remove rows and columns i and j from the matrix
        add a new row and column for the junction (distances from the step above)
        remove labels[i] and labels[j] from the labels list
        add new_label to the labels list in the same position as the new row
        n = n-1
    final = f"({labels[0]},{labels[1]}:{D[0][1]});"
    return final """

    #work on copies so the original matrix and labels aren't modified
    current_matrix = distance_matrix.copy()
    current_labels = labels.copy()
    n = len(current_labels)

    #tracks how many negative branch lengths were corrected
    negative_count = 0

    #each round joins two nodes into one, shrinking the matrix by 1
    #stop when 2 nodes remain, they are connected after the loop
    while n > 2:
        #reset every round because D* values change every round the matrix shrinks
        smallest_so_far = np.inf

        #row sums: how far each node is from all others
        r = current_matrix.sum(axis=1)

        #D* (neighbor-joining matrix), see README: Formulas
        #corrects each distance for how far both nodes are from everyone so nodes that are 
        #only close because they're close to everything aren't joined by mistake
        #only pairs with j>i are checked so a node is never paired with itself
        D_star = np.zeros((n,n))
        for i in range(n):
            for j in range(i + 1, n):
                D_star[i][j] = (n-2) * current_matrix[i][j] - r[i] - r[j]
                #keep the pair with the smallest D* (in case of tie chooses the first one found)
                if D_star[i][j] < smallest_so_far:
                    smallest_so_far = D_star[i][j]
                    best_i = i
                    best_j = j
        
        #branch lengths from the new junction to best_i and best_ 
        #start as an even split of dheir distance and shifted toward the node farther from everyone
        arm_i = (current_matrix[best_i][best_j] + (r[best_i] - r[best_j]) / (n-2)) / 2
        arm_j = current_matrix[best_i][best_j] - arm_i
        
        #if distances aren't perfectly addtive, set the negative arm to 0 and give the other the full distance
        if arm_i < 0:
            negative_count += 1
            arm_i = 0
            arm_j = current_matrix[best_i][best_j]
        if arm_j < 0:
            negative_count += 1
            arm_j = 0
            arm_i = current_matrix[best_i][best_j]

        #distance from the new junction to every other remaining node
        #paths from best_i and best_j to every other node have to be adjusted, add them, subtract the arms, and divide by 2
        new_distances = []
        for other_node in range(n):
            if other_node != best_i and other_node != best_j:
                distance = (current_matrix[best_i][other_node] + current_matrix[best_j][other_node] - current_matrix[best_i][best_j]) / 2
                new_distances.append(distance)

        #newick line; nests inside later labels as joins continue
        new_label = f"({current_labels[best_i]}:{arm_i},{current_labels[best_j]}:{arm_j})"
       
        #indices of the nodes that stay
        keep = []
        for other_node in range(n):
            if other_node != best_i and other_node!= best_j:
                keep.append(other_node)

        #build the smaller (n-1) x (n-1) matrix: kept nodes first, junction last
        new_matrix = np.zeros((n-1, n-1))
        for new_row in range(len(keep)):
            for new_col in range(len(keep)):
                new_matrix[new_row][new_col] = current_matrix[keep[new_row]][keep[new_col]]
        #junction goes in the last row and column (n-2 in the new matrix)
        for new_row in range(len(keep)):
            new_matrix[new_row][n-2] = new_distances[new_row]
            new_matrix[n-2][new_row] = new_distances[new_row]

        #labels in the same order as the new matrix rows    
        new_labels = []
        for old_row in keep:
            new_labels.append(current_labels[old_row])
        new_labels.append(new_label)

        #move on to the next round with a smaller table
        current_matrix = new_matrix
        current_labels = new_labels
        n = n-1
    print(f"Negative branch lengths set to 0: {negative_count}")

    #two nodes left: connect them with one branch of their remaining distance
    final = f"({current_labels[0]},{current_labels[1]}:{current_matrix[0][1]});"
    return final
````

````python
def plot_tree(newick_tree: str) -> None:
    """Plots a phylogenetic tree from a Newick string using Biopython and Matplotlib.

    The tree is rendered in a rectangular layout with smaller leaf labels and
    added margins to reduce overlap between labels and branches.

    Args:
        newick_tree (str): Tree in Newick format

    Returns:
        None: Displays the plotted tree
    """
    handle = StringIO(newick_tree)
    tree = Phylo.read(handle, "newick")

    fig, ax = plt.subplots(figsize=(8, 10))

    Phylo.draw(
        tree,
        axes=ax,
        do_show=False,
        label_func=lambda clade: clade.name if clade.is_terminal() else None,
    )
    for text in ax.texts:
        text.set_fontsize(6)              # shrink labels
    ax.margins(x=0.1, y=0.05)             # extra padding around tree

    plt.tight_layout()
    plt.show()
````
